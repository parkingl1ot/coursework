```python
def h(x0, t, r):
    xt = []
    for j in t:
        x = x0
        for i in range(t[j]):
            x = r * x * (1 - x )
        xt = xt + [x,]  
    return xt


t = np.arange(0, 51)

ya = h(0.01, t, 3.4)
yb = h(0.1, t, 3.5)
yc = h(0.5, t, 3.58)

plt.plot(t, ya)
plt.show()

plt.plot(t, yb)
plt.show()

plt.plot(t, yc)
plt.show()
```
